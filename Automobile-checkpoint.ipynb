```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
# Count the occurrences of each insurance category
category_counts = df['LIBEBRAN'].value_counts()

# Create a pie chart with improved styling
plt.figure(figsize=(10, 7))
plt.pie(category_counts, labels=None, autopct='%1.1f%%', startangle=90, 
        shadow=True, explode=[0.05]*len(category_counts), 
        colors=sns.color_palette('pastel'))

# Add a title
plt.title('Distribution of Insurance Types', fontsize=16, pad=20)

# Add a legend/key outside the pie chart for better readability
plt.legend(category_counts.index, title="Insurance Categories", 
           loc="center left", bbox_to_anchor=(1, 0.5))

# Ensure the pie chart is a circle
plt.axis('equal')

# Add some padding around the chart
plt.tight_layout()

# Display the chart
plt.show()
plt.savefig('insurance_distribution.png', dpi=300, bbox_inches='tight')
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Take only the first 3000 rows from your dataframe 
df_subset = df.head(3000)  # Fixed variable name: df_subnet → df_subset

# Option 1: Use top 20 customers from the subset
top_customers = df_subset['EMIS_PAR'].value_counts().nlargest(20).index
filtered_df = df_subset[df_subset['EMIS_PAR'].isin(top_customers)]

# Create the heatmap using the filtered data
heatmap_data = pd.crosstab(filtered_df['EMIS_PAR'], filtered_df['CATEGORIE_MÈRE'])

# Create the heatmap
plt.figure(figsize=(12, 10))
sns.heatmap(
    heatmap_data, 
    annot=True,  # Show values in cells
    fmt='d',     # Use integer format for annotations
    cmap='YlGnBu',  # Color palette (Yellow-Green-Blue)
    linewidths=0.5,
    cbar_kws={'label': 'Count'}
)

# Add labels and title
plt.title('Customer Fleet Distribution Heatmap (Top 20 Customers)', fontsize=16, pad=20)
plt.xlabel('Fleet Category (CATEGORIE_MÈRE)', fontsize=12)
plt.ylabel('Customer (EMIS_PAR)', fontsize=12)

# Rotate axis labels if needed
plt.xticks(rotation=45, ha='right')

# Improve layout for better readability
plt.tight_layout()

# Show the plot
plt.show()
plt.savefig('Fleet distribution.png', dpi=300, bbox_inches='tight')
```

```python
# Load the dataset
df = pd.read_csv("Automobile_Insurance.csv", low_memory = False)

# Display first few rows
df.head()
```

```python
print(df.shape)
```

```python
# List columns to drop 
columns_to_drop = ["DATEEFFE", "ASSURE", "DATEECHE", "DATEMIS","TAUXPART", "CAPITAL", "NUMEPOLI"]  

# Drop columns
df_cleaned = df.drop(columns=columns_to_drop)
#print df
df_cleaned.head()
df_clean = df_cleaned.select_dtypes(exclude=['object'])
df_clean.head()
```

```python
df_clean.columns
```

```python
# Replace all NaN values with 0
df_clean.fillna(0, inplace=True)
df_clean.head(100)
```

```python
#seperate y from x
X = df_clean.drop(columns = 'PRIMTOTA' , axis = 1)
y = df_clean['PRIMTOTA']
y.head()
X.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn import metrics
```

```python
#Splitting the data into training and testing set .
X_train, X_test, Y_train, Y_test = train_test_split(X, y, test_size = 0.2, random_state = 1)
print(X.shape, X_train.shape, X_test.shape)
```

```python
print(X_train.dtypes)  # Check data types of columns
```

```python
#Model training
automobilemodel = LinearRegression()
#fit model
automobilemodel.fit(X_train, Y_train)
```

```python
#Evaluate model
y_pred = automobilemodel.predict(X_train)
y_pred
```

```python
#R squared values
r2_train = metrics.r2_score(Y_train, y_pred)
print('R2 value: ', r2_train)#value ususally between 0 and 1
```

```python
#prediction on testing set
y_test_pred = automobilemodel.predict(X_test)
y_test_pred
```
